```python
#how the tomography settings are encoded in the raw circuits
import json
from pathlib import Path
from qiskit import qpy
```

```python
data_dir = Path("../../data/bell_2q_decoherence/state 00+11")

with open(data_dir / "2_circuitsraw_64.qpy", "rb") as f:
    circuits_raw = qpy.load(f)

with open(data_dir / "2_counts_64.json", "r", encoding="utf-8") as f:
    counts = json.load(f)

print("raw circuits:", len(circuits_raw))
print("counts entries:", len(counts))
```

```python
def time_label(i):
    if 0 <= i <= 15:
        return "t0_0dt"
    elif 16 <= i <= 31:
        return "t1_4000dt"
    elif 32 <= i <= 47:
        return "t2_8000dt"
    elif 48 <= i <= 63:
        return "t3_16000dt"
    return "unknown"
```

```python
for i in range(16):
    print(f"index={i}, time={time_label(i)}, ops={circuits_raw[i].count_ops()}")
```

```python
for i in range(16):
    print(f"\n=== basis_index {i} ===")
    print(circuits_raw[i].draw("text"))
```

```python
for i in range(16):
    qc = circuits_raw[i]
    print(f"\n=== basis_index {i} ===")
    for instr in qc.data:
        op = instr.operation.name
        qubits = [q._index for q in instr.qubits]
        print(op, qubits)
```

```python
for i in range(16):
    qc = circuits_raw[i]
    print(f"\n=== basis_index {i} ===")
    for instr in qc.data:
        op = instr.operation.name
        qubits = [q._index for q in instr.qubits]
        if op not in ["h", "cx", "barrier", "measure"]:
            print(op, qubits)
```

```python
def extract_basis_gates(qc):
    basis_gates = []
    for instr in qc.data:
        op = instr.operation.name
        qubits = [q._index for q in instr.qubits]
        if op not in ["h", "cx", "barrier", "measure"]:
            basis_gates.append((op, tuple(qubits)))
    return basis_gates
```

```python
basis_map = {}

for i in range(16):
    basis_map[i] = extract_basis_gates(circuits_raw[i])

for k, v in basis_map.items():
    print(f"basis_index {k}: {v}")
```

```python
patterns = {}
for i in range(16):
    pattern = tuple(basis_map[i])
    patterns.setdefault(pattern, []).append(i)

for pattern, indices in patterns.items():
    print(f"indices {indices}: {pattern}")
```
